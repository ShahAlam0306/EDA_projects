5) 10 best bowling figures

```python
df['highest_wickest'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
df['highest_wickest'] = df['highest_wickest'].astype(int)
top_bowlers = df.groupby('best_bowling')['highest_wickest'].sum().sort_values(ascending = False)
top_bowlers.head(5)
```

```python
top_bowlers.plot(kind = 'bar')
```
